```python
#07  Function: grab the domain from "user@domain.com".

def domain(email):
    return email.split('@')[1]

print(domain('user@domain.com'))
```

```python
#08  Function: return True if the word "dog" is in a string (case-insensitive)

def containe_dog(text):
    return 'dog' in text.lower()
print(containe_dog('Is there a dog here?'))
print(containe_dog('Is there a cat here?'))
    
```

```python
#09 Function: count how many times "dog" occurs in a string (case-insensitive).
def count_dog(text):
    return text.count('dog')
print(count_dog('This dog runs faster than the other Dog dude!'))
```

```python
#10 caught_speeding(speed, is_birthday): ≤ 60 "No Ticket", 61–80 "Small Ticket", ≥ 81 "Big Ticket". On a birthday, allow 5 mph more.
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed -= 5
    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print(caught_speeding(65, False)) 
print(caught_speeding(65, True))  
print(caught_speeding(85, True))  
print(caught_speeding(85, False)) 
```
